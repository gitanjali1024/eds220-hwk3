````python
# FINAL CODE

# Import libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# Import spills data
fp = os.path.join('data', 'Spill_incidents_20261102.csv')
spills = pd.read_csv(fp)

# Import shapefile
fp = os.path.join('data', 'tl_2026_us_county','tl_2026_us_county.shp')
counties = gpd.read_file(fp)

# Update column names to snake case
spills.columns = spills.columns.str.lower().str.replace(' ', '_')

# Update date columns to datetime format
spills['spill_date'] = pd.to_datetime(spills['spill_date'])
spills['received_date'] = pd.to_datetime(spills['received_date'])
spills['close_date'] = pd.to_datetime(spills['close_date'])

# Filter to petroleum spills in 2023
petr_23 = spills[(spills['spill_date'] >= '2023-01-01') & (spills['spill_date'] <= '2023-12-31') & (spills['material_family'] == 'Petroleum')]

# Create new data frame for spills per county
spills_per_county = petr_23.groupby('county')['spill_number'].nunique().rename('n_spills').reset_index()

# Update column names
counties.columns = counties.columns.str.lower()

# Select ny counties
ny_counties = counties[counties['statefp'] == '36'].copy()

# Update St Lawrence to St. Lawrence
spills_per_county.replace({'St Lawrence' : 'St. Lawrence'}, inplace = True)

# Rename the county column in ny_counties
ny_counties.rename(columns = {'name' : 'county'}, inplace = True)

# Perform an inner join on ny_counties and spills_per_county
ny_spills = pd.merge(ny_counties, spills_per_county, on = 'county', how = 'inner').copy()

# Find number of spills in Tuscrarora and Onondaga Reservation
tuscarora = spills_per_county.loc[spills_per_county['county'] == 'Tuscarora Indian Reservation', 'n_spills'].values[0]
onondaga = spills_per_county.loc[spills_per_county['county'] == 'Onondaga Indian Reservation', 'n_spills'].values[0]

# Add Tuscarora and Onondaga spills to data frame
ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'] = ny_spills.loc[ny_spills['county'] == 'Niagara', 'n_spills'] + tuscarora
ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'] = ny_spills.loc[ny_spills['county'] == 'Onondaga', 'n_spills'] + onondaga

# Plot the data
fig, ax = plt.subplots(1, 1, figsize=(12, 10))

ny_spills.plot(
    column="n_spills",
    cmap="YlOrRd",
    linewidth=0.6,
    ax=ax,
    edgecolor="dimgray",
    legend=True,
    legend_kwds={
        "label": "Number of Petroleum Spills",
        "orientation": "horizontal",
        "shrink": 0.6,
        "pad": 0.05,
    },
)

ax.set_title(
    "Petroleum Spills by County\nNew York State (2023)",
    fontsize=16,
    fontweight="bold",
    pad=20,
)

ax.set_axis_off()

ax.text(
    x=0.02,
    y=0.02,
    s=(
        "Data Source: New York State Department of Environmental Conservation\n"
        "Date of Access: November 2, 2026"
    ),
    transform=ax.transAxes,
    fontsize=9,
    style="italic",
    color="darkgray",
    verticalalignment="bottom",
    horizontalalignment="left",
)

plt.tight_layout()
plt.show()
````
